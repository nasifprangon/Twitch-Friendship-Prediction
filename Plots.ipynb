```python
import matplotlib.pyplot as plt
import numpy as np

models = ['Logistic Regression', 'XGBoost', 'SVM']
accuracy_without = [0.69, 0.69, 0.63]
accuracy_with = [0.88, 0.92, 0.81]

x = np.arange(len(models))
width = 0.35

plt.figure(figsize=(8, 5))
plt.bar(x - width/2, accuracy_without, width, label='Without Graph', color='#1f77b4')  # blue
plt.bar(x + width/2, accuracy_with, width, label='With Graph', color='#ff7f0e')       # orange

plt.ylabel('Accuracy')
plt.title('Accuracy Comparison by Model')
plt.xticks(x, models, rotation=15)
plt.ylim(0.5, 1.0)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
auc_without = [0.7671, 0.7536, 0.6800]
auc_with = [0.9351, 0.9688, 0.9360]

plt.figure(figsize=(8, 5))
plt.bar(x - width/2, auc_without, width, label='Without Graph', color='#1f77b4')  # blue
plt.bar(x + width/2, auc_with, width, label='With Graph', color='#ff7f0e')       # orange

plt.ylabel('AUC')
plt.title('AUC Comparison by Model')
plt.xticks(x, models, rotation=15)
plt.ylim(0.6, 1.0)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
methods = ['Node2Vec\nRandom', 'Node2Vec\nAdamic-Adar Priority', 'Node2Vec\nCommon Neighbor Priority']
accuracy = [0.9465, 0.87, 0.9082]
f1_score = [0.9464, 0.87, 0.9081]
auc = [0.9889, 0.9429, 0.9696]

x = np.arange(len(methods))  # label locations
width = 0.25  # width of the bars

# Plot
fig, ax = plt.subplots(figsize=(10, 5))
bars1 = ax.bar(x - width, accuracy, width, label='Accuracy', color='steelblue')
bars2 = ax.bar(x, f1_score, width, label='F1 Score', color='darkorange')
bars3 = ax.bar(x + width, auc, width, label='AUC', color='mediumseagreen')

# Labels
ax.set_ylabel('Score')
ax.set_title('📊 Node2Vec Model Comparison')
ax.set_xticks(x)
ax.set_xticklabels(methods)
ax.set_ylim(0.85, 1.0)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
models = ['Node2Vec', 'Node2Vec+Attributes', 'Node2Vec+Attributes+Adamic-Adar']
accuracy = [0.9585, 0.9588, 0.9691]
f1 = [0.9592, 0.9594, 0.9694]
auc = [0.9920, 0.9919, 0.9944]

x = np.arange(len(models))
width = 0.25

# Plot
plt.figure(figsize=(10, 6))
plt.bar(x - width, accuracy, width, label='Accuracy', color='royalblue')
plt.bar(x, f1, width, label='F1-score', color='darkorange')
plt.bar(x + width, auc, width, label='AUC', color='forestgreen')

plt.xlabel('Model Variant', fontsize=13)
plt.ylabel('Score', fontsize=13)
plt.title('Comparison of Node2Vec Variants', fontsize=14)
plt.xticks(x, models)
plt.ylim(0.94, 1.00)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
